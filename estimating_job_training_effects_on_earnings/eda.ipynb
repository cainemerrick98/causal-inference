## Structure of the repo

```python
import dowhy.datasets
import matplotlib.pyplot as plt
```

```python
data = dowhy.datasets.lalonde_dataset()
data.head()
```

```python
treated = data[data['treat'] == 1]
control = data[data['treat'] == 0]
```

# EDA

## Why and what have we found

```python
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111)

ax.hist(treated['age'], bins=20, alpha=1, label='Treated')
ax.hist(control['age'], bins=20, alpha=0.5, label='Control')
ax.set_xlabel('Age')
ax.set_ylabel('Frequency')
ax.set_title('Distribution of Age by Treatment Status')
ax.legend()
```

```python
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111)

ax.hist(treated['educ'], bins=20, alpha=1, label='Treated')
ax.hist(control['educ'], bins=20, alpha=0.5, label='Control')
ax.set_xlabel('Education')
ax.set_ylabel('Frequency')
ax.set_title('Distribution of Age by Treatment Status')
ax.legend()
```

```python
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111)

ax.bar('Treated', treated['married'].value_counts(normalize=True).loc[1], alpha=1, label='Treated')
ax.bar('Control', control['married'].value_counts(normalize=True).loc[1], alpha=0.5, label='Control')
ax.set_xlabel('Married')
ax.set_ylabel('Proportion')
ax.set_title('Proportion of Married by Treatment Status')
ax.legend()
```
